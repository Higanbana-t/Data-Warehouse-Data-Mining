```python
# ==============================================================================
# NOTEBOOK 04: TRÌNH DIỄN KẾT QUẢ HUẤN LUYỆN MÔ HÌNH HỌC MÁY
# ==============================================================================
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image, display
from pathlib import Path
import os

# Trỏ về thư mục gốc dự án
if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')

tables_dir = Path("reports/tables")
figs_dir = Path("reports/figures")

print("✅ Sẵn sàng đọc kết quả đánh giá từ thư mục reports/")
```

```python
print("🎯 1. BÀI TOÁN NHỊ PHÂN: PHÂN BIỆT TRAFFIC HỢP LỆ VÀ TẤN CÔNG")
csv_binary = tables_dir / "model_comparison_binary.csv"

if csv_binary.exists():
    df_bin = pd.read_csv(csv_binary)
    display(df_bin)
    
    # Hiển thị biểu đồ so sánh
    img_bin = figs_dir / "model_comparison_binary.png"
    if img_bin.exists():
        display(Image(filename=str(img_bin), width=800))
    
    # Hiển thị Ma trận nhầm lẫn của Random Forest (mô hình thường tốt nhất)
    img_cm = figs_dir / "confusion_binary_rf.png"
    if img_cm.exists():
        print("\nMa trận nhầm lẫn (Confusion Matrix) - Random Forest:")
        display(Image(filename=str(img_cm), width=500))
else:
    print("⚠️ Chưa có kết quả. Hãy chạy lệnh `python -m src.mining.train` trước.")
```

```python
print("🎯 2. BÀI TOÁN ĐA LỚP: NHẬN DIỆN CHI TIẾT NHÓM TẤN CÔNG")
csv_multi = tables_dir / "model_comparison_multiclass.csv"

if csv_multi.exists():
    df_multi = pd.read_csv(csv_multi)
    display(df_multi)
    
    img_multi = figs_dir / "model_comparison_multiclass.png"
    if img_multi.exists():
        display(Image(filename=str(img_multi), width=800))
else:
    print("⚠️ Chưa có kết quả đa lớp.")
```
